### Text reuse policy

AI wording suggestions were not copied blindly. They were reviewed, corrected, shortened, and adapted by the student before use.
- Where AI-generated English wording was accepted with only minor edits after student review → **`direct_text_reused: true`**
- Where AI was used only for structure/ideas and the student rewrote the content → **`direct_text_reused: false`**
- Where AI output was rejected due to factual errors → **`direct_text_reused: false`**

## M4-004 — AI error caught: wrong schedulability inequality ⚠️

**Student prompt:**
> *"In the formal model section, write the schedulability condition and explain it"*

**Claude's error:** Wrote $U_p < 1$ (strict inequality).

**Why wrong:** From my M1 notes and Buttazzo Chapter 6, the correct EDF schedulability condition is $U_p \leq 1$ (non-strict). Under EDF, $U_p = 1$ is **still schedulable** — this is EDF's key advantage over RM. The strict inequality misrepresents the theorem.

**What I did:** Caught this from my own M1 knowledge and Buttazzo p. 239. Told Claude:
> *"That is wrong. It should be $U_p \leq 1$, not $U_p < 1$. Under EDF a system at 100% utilisation is still schedulable. Please correct."*

Claude corrected it. Final paper uses the correct $\leq$.

**`direct_text_reused: false`** — The erroneous text was rejected. The correction came from my own M1 knowledge.

## M4-008 — AI error caught: fabricated statistic ⚠️

**Student prompt:**
> *"Write the concrete strengths and limitations sections from M3…"*

**Claude's error:** First draft included:
*"In typical workloads IPE reduces mean aperiodic response time by approximately 30% compared to DPE."*

**Why wrong:** This figure does not appear in Buttazzo Chapter 6, Spuri & Buttazzo (1996), or anywhere in my M1–M3 work. It is a fabricated number. Including it would be academic dishonesty — I cannot cite a statistic with no traceable source.

**What I did:**
> *"That 30% figure is not in my reference. Do not invent numbers. Replace it with a qualitative statement only."*

Claude revised to: *"IPE serves aperiodic requests almost as fast as the optimal EDL server and always at the highest system priority when capacity is available."* This is qualitative and matches Buttazzo.

**`direct_text_reused: true`** — The fabricated sentence was rejected. The corrected remainder of the section was reviewed and used. The section is AI-assisted English of my M3 evaluation, with one erroneous sentence removed and replaced.

## Statistical overview
